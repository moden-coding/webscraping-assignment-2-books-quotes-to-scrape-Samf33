```python
url = "https://books.toscrape.com/catalogue/category/books_1/index.html"
resp = requests.get(url, timeout=20)
soup = BeautifulSoup(resp.text, "html.parser")
titles = soup.find_all('h3')
better_titles = [ti.find('a').get('title') for ti in titles]
stars = soup.find_all('p', class_ = "star-rating")
better_stars = list(map(lambda l: l.get('class')[1], stars))
prices = soup.find_all('p', class_ = "price_color")
better_prices = [p.get_text().strip('Â£') for p in prices]
zipped = list(zip(better_prices,better_stars))
nums = {"One":1, "Two": 2, "Three":3, "Four": 4, "Five":5}
new_combination = []
for l in zipped:
    l = (l[0],nums[l[1]])
    new_combination.append(l)
ratios = {}
for price, star in new_combination:
    if star in ratios:
        ratios[star]["count"] += 1
        ratios[star]["avg_price"] = ((ratios[star]["avg_price"] * (ratios[star]["count"] - 1)) +float(price))/ratios[star]["count"]
    else:
        ratios[star] = {"avg_price": float(price), "count": 1}
for fi in ratios.values():
    fi["avg_price"] = round(fi["avg_price"], 2)
sorted_ratios = sorted(ratios.items())
print(sorted_ratios)
```
